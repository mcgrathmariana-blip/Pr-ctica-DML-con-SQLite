```python
import sqlite3

# Crear la base de datos
conexion = sqlite3.connect("practica_dml.db")

# Crear un cursor para ejecutar SQL
cursor = conexion.cursor()

# Crear la tabla vendedores
cursor.execute("""
CREATE TABLE IF NOT EXISTS vendedores (
    id INTEGER PRIMARY KEY,
    nombre TEXT NOT NULL,
    ventas_totales REAL,
    zona TEXT,
    fecha_ingreso TEXT
)
""")

conexion.commit()

print("Base de datos y tabla creadas correctamente.")
```

```python
# Insertar vendedores iniciales
vendedores = [
    (1, 'Ana López', 8500, 'Norte', '2024-01-15'),
    (2, 'Carlos Pérez', 12000, 'Sur', '2023-06-10'),
    (3, 'Lucía Gómez', 9500, 'Norte', '2024-03-20'),
    (4, 'Martín Díaz', 15000, 'Este', '2022-11-05'),
    (5, 'Sofía Torres', 7000, 'Oeste', '2024-05-12')
]

cursor.executemany("""
INSERT INTO vendedores
(id, nombre, ventas_totales, zona, fecha_ingreso)
VALUES (?, ?, ?, ?, ?)
""", vendedores)

conexion.commit()

print("Vendedores iniciales insertados correctamente.")
```

```python
# Consultar vendedores de la zona Norte
cursor.execute("""
SELECT nombre, ventas_totales
FROM vendedores
WHERE zona = 'Norte'
""")

resultados = cursor.fetchall()

for vendedor in resultados:
    print(vendedor)
```

```python
# Insertar dos nuevos vendedores
cursor.execute("""
INSERT INTO vendedores
(id, nombre, ventas_totales, zona, fecha_ingreso)
VALUES
(6, 'Diego Fernández', 11000, 'Norte', '2025-01-10')
""")

cursor.execute("""
INSERT INTO vendedores
(id, nombre, ventas_totales, zona, fecha_ingreso)
VALUES
(7, 'Valentina Ruiz', 9000, 'Sur', '2025-02-15')
""")

conexion.commit()

print("Los dos nuevos vendedores fueron insertados correctamente.")
```

```python
# Insertar tres vendedores en una sola sentencia
cursor.executemany("""
INSERT INTO vendedores
(id, nombre, ventas_totales, zona, fecha_ingreso)
VALUES (?, ?, ?, ?, ?)
""", [
    (8, 'Pedro Sánchez', 13000, 'Sur', '2025-03-01'),
    (9, 'Camila Herrera', 7500, 'Norte', '2025-03-10'),
    (10, 'Julián Castro', 16000, 'Este', '2025-03-15')
])

conexion.commit()

print("INSERT masivo realizado correctamente.")
```

```python
# Actualizar las ventas del vendedor con ID 2
cursor.execute("""
UPDATE vendedores
SET ventas_totales = ventas_totales + 500
WHERE id = 2
""")

conexion.commit()

print("Ventas del vendedor actualizadas correctamente.")
```

```python
# Verificar la actualización
cursor.execute("""
SELECT id, nombre, ventas_totales
FROM vendedores
WHERE id = 2
""")

print(cursor.fetchone())
```

```python
# Cambiar la zona de los vendedores con más de 10.000 en ventas
cursor.execute("""
UPDATE vendedores
SET zona = 'Internacional'
WHERE ventas_totales > 10000
""")

conexion.commit()

print("Zonas actualizadas correctamente.")
```

```python
cursor.execute("""
SELECT id, nombre, ventas_totales, zona
FROM vendedores
WHERE ventas_totales > 10000
""")

for fila in cursor.fetchall():
    print(fila)
```

```python
# Eliminar el vendedor de prueba
cursor.execute("""
DELETE FROM vendedores
WHERE id = 10
""")

conexion.commit()

print("Vendedor de prueba eliminado correctamente.")
```

```python
cursor.execute("""
SELECT *
FROM vendedores
WHERE id = 10
""")

print(cursor.fetchall())
```
